```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
# load dataset 
phool= sns.load_dataset('iris')
phool
```

```python
phool.info()
```

```python
phool.columns
```

```python
phool.isnull().sum()
```

```python
phool.species.value_counts().plot(kind='bar')
```

```python
sns.countplot(x='species',hue='sepal_width',data=phool)
plt.title('Barplot',fontsize=14,fontweight='bold')
plt.xlabel('species of iris',fontsize=12)
plt.ylabel('count of species',fontsize=12)
plt.show()
```

```python
# scatterplot
# plt.figure(figsize=(10,10))
sns.scatterplot(x='petal_length',y='petal_width', hue='species',data=phool,palette='Set2')
plt.title('Phool Data: Petal Length vs Petal Width',fontsize=14,fontweight='bold')
plt.xlabel('Length of Petal (cm)',fontsize=12)
plt.ylabel('Width of Petal (cm)',fontsize=12)
```

```python
# Boxplot
plt.figure(figsize=(8,5))
sns.boxenplot(x='species',y='sepal_length',data=phool,palette='plasma')
plt.title('sepal length distribution by species')
plt.xlabel('Flower Species Type')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
# Heatmap plot
plt.figure(figsize=(7,5))
corr=phool.corr(numeric_only=True)
sns.heatmap(corr)
plt.title('Correlation Matrix of Phool Measurements')
```

```python
# Pairplot
sns.pairplot(data=phool,hue='species', palette='Set1')
```

```python
# Histogram
phool.hist(bins=15,grid=False)
```

```python
# Stripplot
sns.stripplot(x='species', y='petal_length', data=phool, hue='species')
 
```
