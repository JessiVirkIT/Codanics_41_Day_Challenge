```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dataset
kashti = sns.load_dataset('titanic')
kashti
```

```python
kashti.info()
```

```python
kashti.isnull().sum()
```

```python
kashti.survived.value_counts().plot(kind='bar')
```

```python
sns.countplot(x ='survived' , hue = 'sex' , data = kashti)
```

```python
# scatter plot
plt.scatter(x ='age' , y ='fare', data = kashti)
```

```python
sns.scatterplot(x = 'age' , y = 'fare' , hue='pclass', data = kashti )
plt.title('Titanic Passenger Ticket Cost vs Passenger Age',fontsize=14,fontweight='bold')
plt.xlabel('Age of Passengers (Years)',fontsize=12)
plt.ylabel('Ticket Fare Paid (Pounds)',fontsize=12)
plt.show()
```

```python
sns.barplot(x='sex',y='fare',hue='pclass',data=kashti)
```

```python
kashti.columns
```

```python
# Pairplot
sns.pairplot(kashti)
```

```python
#  Histogram plot
kashti.hist()
```

```python
#  bars ko bins boltay hein
kashti.hist(bins=10 , figsize=(9,7), grid=False)
```

```python
# boxplot
sns.boxplot(x='sex',y='fare',hue='pclass',data=kashti,palette='Set3')
```

```python
sns.boxenplot(x='sex',y='fare',hue='pclass',data=kashti,palette='plasma')
```

```python
#  Assignment: how to change  x and y axis titles?
```

```python
g = sns.FacetGrid(kashti, hue='survived',col='pclass', margin_titles=True,
palette={1:'seagreen',0:'gray'})
g=g.map(plt.scatter,'fare','age',edgecolor='w').add_legend();
```

```python
g = sns.FacetGrid(kashti, hue='survived',col='who', margin_titles=True,
palette={1:'seagreen',0:'gray'})
g=g.map(plt.scatter,'fare','age',edgecolor='w').add_legend();
```

```python
sns.stripplot(x='sex',y='fare',hue='pclass',data=kashti)
```

```python
# heatmap
corr= kashti.corr(numeric_only=True) # correlation matlb kisi cheez ka barhna ya kam hona ek chez ki wjah say
plt.figure(figsize=(10,10))
sns.heatmap(corr)
plt.title('Heat map of kashti')
```

```python
sns.violinplot(x='sex',y='fare',data= kashti,hue='class')
```
