```python
# Average final grade
df['G3'].mean()
```

```python
# maximum average
df['G3'].max()
```

```python
# minimum average
df['G3'].min()
```

```python
# frequent final garade count values
df['G3'].value_counts().sort_index()
```

```python
# gender categories
df['sex'].value_counts()
```

```python
# Gender-wise average grades
df.groupby('sex')['G3'].mean()
```

```python
# school average final grade
df.groupby('school')['G3'].mean()
```

```python
# different study time categories average final grade
df.groupby('studytime')['G3'].mean()
```

```python
# past class failures average final grade
df.groupby('failures')['G3'].mean()
```

```python
# top 10 students 
df.sort_values(by='G3',ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt

df.groupby('studytime')['G3'].mean().plot(kind='bar')
plt.title('Study Time vs Average Final Grade')
plt.xlabel('Study Time Category')
plt.ylabel('Average Final Grade')
plt.tight_layout()
plt.show()
```

```python
df['G3'].plot(kind='hist', bins=10)
plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

## Key Findings

1. The dataset contains 395 students and 33 columns.
2. The average final grade (G3) is 10.42 out of 20.
3. The highest final grade is 20, and the lowest is 0.
4. The average final grade is higher for male students (10.91) than for female students (9.97) in this dataset.
5. Study-time category 3 has the highest average final grade (11.40).
6. Students with zero past class failures have an average final grade of 11.25, while students with three past failures have an average of 5.69.
7. No missing values or duplicate rows were found in the dataset.
