```python
import pandas as pd
import numpy as np

# Seed for reproducibility
np.random.seed(42)

# Generate Synthetic Professional Dataset
n_rows = 500

data = {
    'Transaction_ID': [f'TXN-{1000+i}' for i in range(n_rows)],
    'Customer_ID': np.random.choice([f'CUST-{i:03d}' for i in range(1, 50)], size=n_rows),
    'Order_Date': pd.date_range(start='2023-01-01', periods=n_rows, freq='h').strftime('%Y-%m-%d'),
    'Product_Category': np.random.choice(['Electronics', 'Clothing', 'Home & Kitchen', 'Books', 'Toys', None], size=n_rows, p=[0.3, 0.25, 0.2, 0.1, 0.1, 0.05]),
    'Price_Per_Unit': np.random.choice([15.0, 25.5, 120.0, 450.0, 1200.0, np.nan, -50.0], size=n_rows),
    'Quantity': np.random.choice([1, 2, 3, 4, 5, 10, 0, -1], size=n_rows),
    'Payment_Method': np.random.choice(['Credit Card', 'PayPal', 'Cash on Delivery', 'Crypto', 'credit_card'], size=n_rows),
    'Customer_Rating': np.random.choice([1, 2, 3, 4, 5, np.nan, 10], size=n_rows), # 10 is an outlier/error
    'Country': np.random.choice(['Pakistan', 'USA', 'UK', 'UAE', 'pakistan', 'US', None], size=n_rows)
}

df = pd.DataFrame(data)

# Add some duplicate rows deliberately
df = pd.concat([df, df.iloc[:15]], ignore_index=True)

# Save to CSV
df.to_csv('professional_practice_data.csv', index=False)
print("Dataset successfully created and saved as 'professional_practice_data.csv'!")
```

# Level 2: Data Cleaning

```python

# 1.Duplicates: Duplicate rows check karnay k liye yay command lgai 
# yahan par khud say socha k agr sum na lgaon tu kya hoga tu, agr sum na lgay tu table mey 
# ek list nazr ay gi jis mey 
# true false likha hoga ,true ka matlb duplicate h or false ka matlb duplicate ni h
# tu sum ka function lgany say hmay basically in saaray True values ko aapas mein ginte 
# kar ke ek final number (jaise 15 ya 0) de deta hai.
df.duplicated().sum()
```

```python
#  ab enhein remove/drop karnay k liye hmnay yay command lgai
# yahn pay hm yay duplicates ko change ni kray ga bas ek temporary table dikha dey ga, 
# esko change ya drop krnay k
# do simple traikay hein matlb do commands hein 
# 1. yay command df.drop_duplicates(inplace=True)
# 2.dosra simple yay k df assign kr dena h e.g; df = df.drop_duplicates(), ma yai wala 
# tareeka kron gi
df = df.drop_duplicates()
```

```python
#  ab yahan pay hmnay fr say chcek kiya k duplicate remove ya drop howay ya ni 
df.duplicated().sum()
```

```python
# 2. Inconsistent Categorical Values:eska matlb ek hi category ya nam ko alg alg spelling,
# capital,small ya space 
# ki wjah sya ghalt likhna jesay Pakistan ,pakistan,Credit Card,credit card
df['Payment_Method'] = df['Payment_Method'].replace({'Credit Card':'credit card',
                                                     'Pakistan':'pakist','Crypto':'crypto'})
# (df[...] = ...): Yeh likhna zaroori hai, warna data permanently change nahi hota.
# Yeh 'Credit Card' ko badal kar 'credit card' kar dega.
# {} yay curly brackets hm tab lgay gay jab hmnay replace mey do say zeada cheezein likhni 
# hun tab
```

```python
# (.unique()): Yeh bina print ke aapko screen par saare payment methods ki list dikha dega. 
# Agar list mein credit card (small letter mein) nazar aa jaye, to iska matlab hai change ho 
# gaya hai.
df['Payment_Method'].unique()
# Aakhir mein .unique() lagane se aapko saare badle hue words ek sath nazar aa jayenge
```

```python
# ab yahan pay country colum ko clean krna tha tu hmnay sab say pehly country colum akelay
# ko command lga k dekh liya 
df['Country'].unique()
```

```python
# ab yahan pay bhi hmnay spelling change kiye 
df['Country'] = df['Country'].replace({'Pakistan':'pakistan','USA':'US'})
df['Country'].unique()
```

```python
# ab yay hmnay fr say chcek krnay k liye command lgai
df[['Price_Per_Unit','Quantity']]
```

```python
# 3.Price_Per_Unit aur Quantity mein negative values (-50, -1) ya zero hain. 
# In wrong rows ko filter out karen.
df = df[(df['Price_Per_Unit'] > 0 ) & (df['Quantity'] > 0)]
```

```python
df[['Price_Per_Unit','Quantity']].T
```

```python
# Customer_Rating mein maximum rating 5 honi chahiye, lekin 10 bhi likha hua hai. 
# Un 10 wali invalid ratings ko correct karen ya drop/null karen.
df.loc[df['Customer_Rating'] > 5,['Customer_Rating']] =np.nan
```

```python
df['Customer_Rating'] = df['Customer_Rating'].replace(10,5)
```

```python
df['Customer_Rating'].unique()
```

```python
# 4.Missing Values (Handling NaNs):
# --> Check karen ke kis kis column mein NaN values hain.
df.isnull().sum()
# df.isnull(): Pure data mein jahan NaN hoga use dhoondegi.
# sum(): Har column ke samne total missing values ka number (count) dikha degi.
# Jis column ke samne 0 hoga, matlab us mein koi NaN value nahi hai.
```

```python
# --> Product_Category mein missing values ko 'Unknown' se fill (impute) karen. 
df['Product_Category'] = df['Product_Category'].fillna('Unknown')
#  Code ke shuru mein df['Product_Category'] = is liye lagaya taake purani khali jagah
# ki jagah naya data save (update) ho jaye.
```

```python
df['Product_Category'].head(10)
```

```python
df['Customer_Rating'] = pd.to_numeric(df['Customer_Rating'],errors='coerce')
```

```python
df['Customer_Rating'] = df['Customer_Rating'].fillna(df['Customer_Rating'].median())
```

```python
df['Customer_Rating'].head()
```

```python
df['Country'] = df['Country'].fillna('Unknown')
```

```python
df['Country'].head()
```

```python
# --> Price_Per_Unit mein missing values ko category-wise Median se fill / impute karen.
df['Price_Per_Unit'] = df['Price_Per_Unit'].fillna(df.groupby('Product_Category')
                                                   ['Price_Per_Unit'].transform('median'))
```

```python
df[['Product_Category','Price_Per_Unit']].head(10)
```

# Level 3: Feature Engineering & Analysis (Analysis & Naye Columns)
- Analysis & Reports (Shuru Karte Hain!)
- Safaai (Cleaning) ho gayi! Ab hum Company Owner ko reports bana kar denge. Iske 3 main tasks hain:

```python
# 1.Total Sales Column: Ek naya column banaen Total_Amount = Price_Per_Unit * Quantity.
# matlb k ;Total Sales Ka Column Banaen
# Price aur Quantity ko multiply karke batayein ke har order kitne ka tha.
df['Total_Amount'] = df['Price_Per_Unit'] * df['Quantity']
# Kyun kar rahe hain? Owner ko yeh dekhna hai ke ek transaction se kul kitne paise aaye
```
