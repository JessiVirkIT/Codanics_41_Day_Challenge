# Steps to Analyze the EDA

1. **Data Understanding:** Data ka basic structure, rows/columns ki tadad (shape), aur data types (numerical/categorical) check karna.
2. **Data Cleaning:** Khali jagah (missing values) ko fill ya remove karna, duplicates khatam karna, aur ghalat data ko durust karna.
3. **Descriptive Statistics:** Data ka mathematical jaiza lena, jaise Mean, Median, Mode, aur Range maloom karna.
4. **Univariate Analysis:** Ek waqt mein sirf ek variable (column) ka jaiza lena (Histogram ya Bar Chart ke zariye).
5. **Bivariate/Multivariate Analysis:** Do ya do se zyada variables ke darmiyan talluq (relationship) aur correlation dekhna (Scatter Plot ya Heatmap ke zariye).
6. **Outlier Detection:** Data mein mojood ajeeb ya bohat zyada alag values (outliers) ko dhoondna aur unhein handle karna (Box Plot ke zariye).
7. F**eature Engineering:** Machine learning ke liye data ko tayyar karna, jaise text ko numbers mein badalna (encoding) aur scaling karna.

# Solution Assignment:
**1. Methods to Find Missing Values**
  - **Count:** df.isnull().sum() (Har column mein missing values ki tadad)
  - **Percentage:** df.isnull().sum() * 100 / len(df) (Missing data ka tasnab/percentage)
  
**2. Handling Missing Values (By Data Type)**
  
  **A. Numeric Data (int, float)**
  - **Mean:** Data normal ho tu Average se fill karein.df['col'].fillna(df['col'].mean(), inplace=True)
  - **Median:** Data mein outliers (extreme values) hon tu Middle value se fill karein.df['col'].fillna(df['col'].median(), inplace=True)
  
**B. Categorical / Object Data (String/Text)**
  - **Mode:** Jo category sab se zyada baar aayi ho (Most Frequent), us se fill karein.df['col'].fillna(df['col'].mode()[0], inplace=True)
  - **New Category:** Missing jagah par 'Unknown' ya 'Missing' likh dein.df['col'].fillna('Unknown', inplace=True)
 
**C. Boolean Data (True / False)**
  -  **Mode:** Jo value (True ya False) zyada baar aayi ho, wohi daal dein.
  - **Logical:** Doosre columns ko dekh kar guess karein (e.g., agar Salary == 0 hai tu Is_Employed = False kar dein).
  
**3. Golden Rule: When to Drop (Delete)?**
  - **Row Drop:** Agar prediction wali main value (Target Variable) missing ho. (df.dropna(subset=['target']))
  - **Column Drop:** Agar column mein 50% se zyada data missing ho. (df.drop(columns=['col']))


# Core Definitions

- **Mean (Average):** The total sum of all numbers divided by the count of numbers.
- **Median (Middle):** The exact middle value when the numbers are sorted from smallest to largest.
- **Mode (Most):** The number that appears most frequently in the set.
- **Range (Difference):** The distance between the highest and lowest values.
