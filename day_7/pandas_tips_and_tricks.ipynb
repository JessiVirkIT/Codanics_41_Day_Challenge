# 16- Splitting a string into multiple columns
- es mey hm dekhay gay k kis trah ek string ko multiple columns k andr convert krna h 
- yay hmary pas tab use hoti h jab ek ya ek say zeada dataframes k tarikay say btai gai hun cheezein for example: neechay code say dekhna
- sab say pehly import krni hein libraries 
  > yay tip boht use hoti h 

```python
# 1. import libraries
import pandas as pd
# 2.second Dataframe ya dataset bnaya , dataframe mey name diya location btai
df = pd.DataFrame({'name':['Yasmin Akhter','Ismat Nawab','Sadia Iqbal','Amna Bibi'],
                  'location':['Islamabad,Pakistan','Danewal,Pakistan','Vehari,Pakistan','Anwaraabad ,PAkistan']})
df              
```

```python
# 3. ab next h k hmnay esko split kesay krna h
# aksr hmary pas data esay ata h lakin jab hm usay utilize krnay ki ksohsi krtay hein
# then we have to actual split data into diferent things
# for example: df dot name, hmnay esko convert krna h phr str h k yay name hmary pas string h 
# fr es k bad str dot split k andr esko split krna h
# convert krnay k liye separation kya honi chae,sepertaion kis cheez say krna h tu ma kehti hn 
# k space say separate krna h matlb yasmin k bad space ay sath .head kr k dekhna h 
df.name.str.split(' ').head()
# es tarikay say hm esko change krtay hein
```

```python
# Splitting  a column into two columns
# ab agr ma lgaon dot expand is equal to true tu expand true krnay k bad kya hoga pora 
# ka pora change ho jay ga 
df.name.str.split(' ',expand=True)
# ab yahan pay hmnay split tu kr diya expand true bhi kr diya
```

```python
# Adding those splits into new columns
# df.name.str.split(' ',expand=True) es code mey changes ni hoi ku , ku k hmnay change kiya hi ni 
# ab hm df es kandr double brackets fr us k andr jo likhay gay us say yay change ho ga
df[['First_Name','Last_Name']]=df.name.str.split(' ',expand=True)
df
# ab es mey change kya ay gi k end par last location thi tu ab end par first name last name ko 
# separation kr k likh diya matlb colum add ho gya
# df.name.str.split(' ',expand=True) jab hmnay esko kiya tu esnay sirf values di lakin 
# jab df[['First_Name','Last_Name']] esko kiya tu yay colum add ho gya
```

```python
# Assignment: location colum ko expand kr k dekhna h 
# yahan pay location hmary pas comma k stah separate ho rai h matlb(',') matlb k location mey 
# islamabad or pakistan k darmiyan comma h 
df[['City','Country']]=df.location.str.split(',',expand=True)
df
# tu es trah hmary pas wai df h lakin ab neechay city or country alg colum mey  agya h
```

```python
# Refine data manipulation
# ab next series mey chltay hein or df ko equal to rakhtay hein df double brackets es k andr jo
# jo colum hm chahty hein wo rakh detya hein
# ma kehti hn pehla column mey yay first_name wala column a jaye or dosra colum last_name 
# a jay or fr teesra city or chotha country  ajay or fr df kr dey gay
df = df[["First_Name","Last_Name","City","Country"]]
df
# ab kya hoga k jo hmnay data manipulate kiya tu usko refine kr k es trah likh liya,
# tu ab hmary pas sirf 4 colum a gaye 
# basically hmnay sab say pehly dataframe bnaya usko split kiya split krnay k bad ek new colum 
# add kiya fr new colum ko select kr liya
# es trah same data say hmary pas es trah ek data set a jay ga
```
