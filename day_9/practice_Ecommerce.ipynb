# Step 1: Python Code Run Karke Dataset Banayein
- Apne Jupyter Notebook ke ek Naye Cell mein yeh code run karein:

```python
import pandas as pd
import numpy as np

np.random.seed(101)
n = 400

hr_data = {
    'Emp_ID': [f'EMP-{200+i}' for i in range(n)],
    'Name': [f'Employee_{i}' for i in range(n)],
    'Department': np.random.choice(['IT', 'Sales', 'HR', 'Finance', 'Marketing', None], size=n, p=[0.3, 0.25, 0.15, 0.15, 0.1, 0.05]),
    'Salary': np.random.choice([45000, 60000, 85000, 120000, 250000, -5000, np.nan], size=n),
    'Age': np.random.choice([22, 28, 35, 42, 55, 120, -5, np.nan], size=n), # 120 & -5 are errors
    'Performance_Score': np.random.choice(['Excellent', 'Good', 'Average', 'Poor', 'good', 'EXCELLENT', None], size=n),
    'Experience_Years': np.random.choice([1, 3, 5, 8, 12, np.nan], size=n),
    'City': np.random.choice(['Karachi', 'Lahore', 'Islamabad', 'karachi', 'ISLAMABAD', None], size=n)
}

df_hr = pd.DataFrame(hr_data)

# Save to CSV
df_hr.to_csv('hr_practice_data.csv', index=False)
print("HR Dataset successfully created as 'hr_practice_data.csv'!")
```

# Step 2: Aap Ke Naye Practice Tasks
- Naye cell mein df = pd.read_csv('hr_practice_data.csv') load karein aur in tasks ko solve karein:

```python
df = pd.read_csv('hr_practice_data.csv')
```

```python
df.head()
```

# 1 Inconsistent Names Clean Karein:
- City column mein 'karachi' aur 'ISLAMABAD' ko standardize karein ('Karachi', 'Islamabad').

```python
df['City'].value_counts()
```

```python
df['City'] = df['City'].replace({'karachi':'karachi' , 'ISLAMABAD':'Islamabad','karachi':'Karachi'})
```

```python
df['City'].value_counts()
```

- Performance_Score mein 'good' aur 'EXCELLENT' ko uppercase/lowercase se clean karke standardize karein.

```python
df['Performance_Score'].value_counts()
```

```python
df['Performance_Score'] = df['Performance_Score'].replace({'good':'Good','EXCELLENT':'Excellent'})
```

```python
df['Performance_Score'].value_counts()
```

# 2 Invalid Age & Salary Handle Karein:
- Age mein negative values (-5) aur wrong age (120) ko filter karke 18 se 65 ke beech wali rows rakhein.

```python
df['Age'].value_counts()
```

```python
df = df[(df['Age'] > 0) & (df['Age'] <= 65)]
```

```python
df['Age'].value_counts()
```

- Salary mein negative values (-5000) ko remove karein.

```python
df['Salary']
```

```python
df= df[(df['Salary'] > 0)] 
```

```python
df['Salary']
```

# 3 Missing Values Impute Karein:
- Department aur City ki missing values ko 'Unknown' se fill karein.

```python
df[['Department','City']]
```

```python
df[['Department','City']] = df[['Department','City']].fillna('Unknown')
```

```python
df[['Department','City']].head(40)
```

- Salary ki missing values ko Department-wise Median se fill karein (groupby('Department')['Salary'].transform('median')).

```python
df['Salary'].head()
```

```python
df['Salary'] = df['Salary'].fillna(df.groupby('Department')['Salary'].transform('median'))
```

```python
df[['Department','Salary']]
```

# 4 Analysis & Bonus Column:
- Ek naya column banayein: Bonus = Salary * 0.10 (10% bonus).

```python
df['Bonus'] = df['Salary'] * 0.10
```

```python
df.head()
```

- GroupBy karke dekhein ke kis Department ki Average Salary sab se zyada hai?

```python
#  Average nikalny k liye k mean() use hota h or sum total nikalny k liye use hota h 
df.groupby('Department')['Salary'].mean()
```

```python
# Agr hm chahtay hein k sab say bari salary wala department sab say opr aye tu agay
# .sor_values lga dein gay
df.groupby('Department')['Salary'].mean().sort_values(ascending=False)
```
