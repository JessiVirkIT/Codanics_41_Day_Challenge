```python
import pandas as pd
import numpy as np

# Seed for reproducibility
np.random.seed(42)

# Generate Synthetic Professional Dataset
n_rows = 500

data = {
    'Transaction_ID': [f'TXN-{1000+i}' for i in range(n_rows)],
    'Customer_ID': np.random.choice([f'CUST-{i:03d}' for i in range(1, 50)], size=n_rows),
    'Order_Date': pd.date_range(start='2023-01-01', periods=n_rows, freq='h').strftime('%Y-%m-%d'),
    'Product_Category': np.random.choice(['Electronics', 'Clothing', 'Home & Kitchen', 'Books', 'Toys', None], size=n_rows, p=[0.3, 0.25, 0.2, 0.1, 0.1, 0.05]),
    'Price_Per_Unit': np.random.choice([15.0, 25.5, 120.0, 450.0, 1200.0, np.nan, -50.0], size=n_rows),
    'Quantity': np.random.choice([1, 2, 3, 4, 5, 10, 0, -1], size=n_rows),
    'Payment_Method': np.random.choice(['Credit Card', 'PayPal', 'Cash on Delivery', 'Crypto', 'credit_card'], size=n_rows),
    'Customer_Rating': np.random.choice([1, 2, 3, 4, 5, np.nan, 10], size=n_rows), # 10 is an outlier/error
    'Country': np.random.choice(['Pakistan', 'USA', 'UK', 'UAE', 'pakistan', 'US', None], size=n_rows)
}

df = pd.DataFrame(data)

# Add some duplicate rows deliberately
df = pd.concat([df, df.iloc[:15]], ignore_index=True)

# Save to CSV
df.to_csv('professional_practice_data.csv', index=False)
print("Dataset successfully created and saved as 'professional_practice_data.csv'!")
```

# Level 2: Data Cleaning(Safaai-Real Proffesional work)

- 1. Duplicates: Duplicate rows chcek karen aur unhein remove/drop karen.

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates().sum()
```

```python
df.duplicated().sum()
```

- 2.Inconsistent Categorical Values :
  - Notice karen k **Payment_Method** mein **'Credit Card'** aur **'credit_card'** dono hein. Unhein ek jaisa standardize/clean karen
  - Country column mein **'Pakistan' aur 'pakistan', 'USA' aur 'US'** hain. Inhein bhi clean karen.

```python
# yay pehly payement_method k colum mey dekha k CRedit Card or credit_card dono hein
#  tu es mey dono hu 
df[['Payment_Method','Country']].value_counts()
```

```python
# ab yahan pay enko clean kiya standardize kiya 
df['Payment_Method'] = df['Payment_Method'].replace({'Credit Card':'credit_card'})
df['Country'] = df['Country'].replace({'Pakistan':'pakistan','USA':'US'})
```

```python
# dubara say chek kiya k clean howa h ya ni 
df[['Payment_Method','Country']].value_counts()
```

 - 3.Invalid/Error Values
   -  **Price_Per_Unit** or **Quantity** mein negative values (-50,-1) ya zero hein.In wrong rows ko filter out karen.
  
   -  **Customer_Rating** mein **maximum rating 5** honi chahiye,**lakin 10** bhi likha howa h. Un 10 wali invalid ratings ko correct karen ya drop/null karen.

```python
df = df[(df['Price_Per_Unit'] > 0) & (df['Quantity'] >0)]
```

```python
df[['Price_Per_Unit','Quantity']]
```

- 4.Missing Values(Handling NaNs)
  - Check karen k **kis colum** mey **NaN values** hein.
  - **Product_Category** mein missing values ko **'Unknown'** se fill(impute) karen.
  - **Price_Per_Unit** mein missing values ko category_wise Median se fill/impute karen. 

```python
df.isnull().sum()
```

```python
df['Product_Category'] = df['Product_Category'].fillna('Unknown')
```

```python
df['Product_Category'].head(40)
```

```python
df['Price_Per_Unit'] = df['Price_Per_Unit'].fillna(df.groupby('Product_Category')['Price_Per_Unit'].transform('median'))
```

```python
df[['Product_Category','Price_Per_Unit']].head()
```
